```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt
data = 100 - np.random.exponential(scale=8, size=1000)
data = np.clip(data, 0, 100)
plt.hist(data, bins=30, edgecolor="black")
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.title("Left-Skewed Distribution")
plt.show()
```
