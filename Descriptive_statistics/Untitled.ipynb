## Descriptive Statistics using numpy#


```python
import numpy as np
from statistics import mode
marks=[45,50,55,60,60,70,75,80,90]
```

```python
print("Mean:",np.mean(marks))
print("Meadian:",np.median(marks))
print("Mode:",mode(marks))
print("Range:",np.ptp(marks))
print("Maximum:",np.max(marks))
print("Minimum:",np.min(marks))
print("Variance:",np.var(marks))
print("Standard Deviation:",np.std(marks))
print("Q1:",np.percentile(marks,25))
print("Q2:",np.percentile(marks,50))
print("Q3:",np.percentile(marks,75))
```
